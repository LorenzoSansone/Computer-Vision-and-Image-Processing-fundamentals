# **Draw with OpenCV**
## Image Processing and Computer Vision


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Line

Syntax:
- cv.line(img, pt1, pt2, color[, thickness[, lineType[, shift]]]) -> img

Parameters:
- **img**: Image.
- **pt1**: First point of the line segment.
- **pt2**: Second point of the line segment.
- **color**: Line color in BGR format (e.g., (0, 255, 0) for green).
- **thickness**: Line thickness.
- **lineType**: Type of the line. See LineTypes.
- **shift**: Number of fractional bits in the point coordinates.


```python
IMG_PATH = './img/ferrari.jpg'
```

```python
##### Read -> Draw BLUE line-> Convert -> Display #####

# 1.1) Read the image as BGR
img_before = cv2.imread(IMG_PATH)

# 1.2) Draw a diagonal blue line with thickness of 5 px
cv2.line(img_before,(0,0),(400,400),(255, 0, 0),5)

# 1.3)Convert the image from BGR to RGB for displaying with matplotlib
img_before = cv2.cvtColor(img_before, cv2.COLOR_BGR2RGB)


##### Read -> Convert -> Draw RED line-> Display #####

# 2.1)Read the image as BGR
img_after1 = cv2.imread(IMG_PATH)

# 2.2) Convert the image from BGR to RGB
img_after1 = cv2.cvtColor(img_after1, cv2.COLOR_BGR2RGB)

# 2.3) Draw a diagonal blue line with thickness of 5 px
cv2.line(img_after1,(0,0),(400,400),(255, 0, 0),5)


##### Read -> Convert -> Draw BLUE line -> Display #####

# 3.1)Read the image as BGR
img_after2 = cv2.imread(IMG_PATH)

# 3.2) Convert the image from BGR to RGB
img_after2 = cv2.cvtColor(img_after2, cv2.COLOR_BGR2RGB)

# 3.3) Draw a diagonal blue line with thickness of 5 px
cv2.line(img_after2,(0,0),(400,400),(0, 0, 255),5)

##################

plt.figure(figsize=(10, 10))

plt.subplot(1, 3, 1)    
plt.imshow(img_before)
plt.title('Line')

plt.subplot(1, 3, 2)    
plt.imshow(img_after1)
plt.title('Line')

plt.subplot(1, 3, 3)    
plt.imshow(img_after2)
plt.title('Line')

plt.show()
```

## Rectangle

Syntax:
- cv.rectangle(img, pt1, pt2, color[, thickness[, lineType[, shift]]]) -> img

Parameters:
- **img**: Image.
- **pt1**: Vertex of the rectangle.
- **pt2**: Vertex of the rectangle opposite to pt1 .
- **color**. Rectangle color or brightness (grayscale image).
- **thickness**: Thickness of lines that make up the rectangle. Negative values, like FILLED, mean that the function has to draw a filled rectangle.
- **lineType**: Type of the line. See LineTypes
- **shift**: Number of fractional bits in the point coordinates.

```python
# Create a black image
img = np.zeros((512,512,3), np.uint8)

cv2.rectangle(img,(20,0),(200,200),(255,0,0),3)

plt.imshow(img)
plt.title("Rectangle")
plt.show()
```

## Circle

Syntax:
- cv.circle(img, center, radius, color[, thickness[, lineType[, shift]]]) -> img

Parameters:
- **img**: Image where the circle is drawn.
- **center**: Center of the circle.
- **radius**: Radius of the circle.
- **color**: Circle color.
- **thickness**: Thickness of the circle outline, if positive. Negative values, like FILLED, mean that a filled circle is to be drawn.
- **lineType**: Type of the circle boundary. See LineTypes
- **shift**: Number of fractional bits in the coordinates of the center and in the radius value.

```python
# Create a black image
img = np.zeros((512,512,3), np.uint8)

cv2.circle(img,(447,63), 100, (0,0,255), -1)

plt.imshow(img)
plt.title("Circle")
plt.show()
```

## Ellipse

Syntax:
- cv.ellipse(	img, center, axes, angle, startAngle, endAngle, color[, thickness[, lineType[, shift]]]	) ->img

Parameters:
- **img**: Image.
- **center**: Center of the ellipse.
- **axes**: Half of the size of the ellipse main axes.
- **angle**: Ellipse rotation angle in degrees.
- **startAngle**: Starting angle of the elliptic arc in degrees.
- **endAngle**: Ending angle of the elliptic arc in degrees.
- **color**: Ellipse color.
- **thickness**: Thickness of the ellipse arc outline, if positive. Otherwise, this indicates that a filled ellipse sector is to be drawn.
- **lineType**: Type of the ellipse boundary. See LineTypes
- **shift**: Number of fractional bits in the coordinates of the center and values of axes.


```python
# Create a black image
img = np.zeros((512,512,3), np.uint8)

cv2.ellipse(img,(256,256),(100,50),0,0,180,255,-1)

plt.imshow(img)
plt.title("Ellipse")
plt.show()
```

## Polygon

Syntax:
- 	cv.polylines(img, pts, isClosed, color[, thickness[, lineType[, shift]]]) -> img

Parameters:
- **img**: Image.
- **pts**: Array of polygonal curves.
- **isClosed**: Flag indicating whether the drawn polylines are closed or not. If they are closed, the function draws a line from the last vertex of each curve to its first vertex.
- **color**: Polyline color.
- **thickness**: Thickness of the polyline edges.
- **lineType**: Type of the line segments. See LineTypes
- **shift**: Number of fractional bits in the vertex coordinates.

```python
# Create a black image
img = np.zeros((512,512,3), np.uint8)
img1 = np.zeros((512,512,3), np.uint8)

pts = np.array([[10,5],[200,300],[70,20],[50,10]], np.int32)
print(pts)

pts = pts.reshape((-1,1,2))
print(pts)

cv2.polylines(img,[pts],True,(0,255,255),3)

cv2.polylines(img1,[pts],False,(0,255,255),3)

plt.figure(figsize=(10, 10))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Polyline closed")

plt.subplot(1, 2, 2)
plt.imshow(img1)
plt.title("Polyline open")

plt.show()
```

## Adding Text to Images

Syntax:
-   cv.putText(	img, text, org, fontFace, fontScale, color[, thickness[, lineType[, bottomLeftOrigin]]]	) ->	img

Parameters:
- **img**: Image.
- **text**: Text string to be drawn.
- **org**: Bottom-left corner of the text string in the image.
- **fontFace**: Font type, see HersheyFonts.
- **fontScale**: Font scale factor that is multiplied by the font-specific base size.
- **color**: Text color.
- **thickness**: Thickness of the lines used to draw a text.
- **lineType**: Line type. See LineTypes
- **bottomLeftOrigin**: When true, the image data origin is at the bottom-left corner. Otherwise, it is at the top-left corner.

```python
# Create a black image
img = np.zeros((512,512,3), np.uint8)

font = cv2.FONT_HERSHEY_SIMPLEX

cv2.putText(img,'OpenCV',(10,500), font, 4,(255,255,255),2,cv2.LINE_AA)

plt.imshow(img)
plt.title("Write text on image")
plt.show()
```
